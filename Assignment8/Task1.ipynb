```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("enriched_students.csv")

df.head()
```

```python
# Step 2: Shape, Info, Dtypes

print("Shape:", df.shape)
print("\nInfo:")
print(df.info())
print("\nData Types:")
print(df.dtypes)
```

```python
# Step 3: Missing values
print("\nMissing Values per Column:")
print(df.isnull().sum())
print("\nPercentage Missing:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
# Step 4: Descriptive statistics
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
# Step 5: Value counts for a categorical column
# Replace 'col' with the name of a categorical column in your dataset

print(df['grade'].value_counts())

print(df['score_category'].value_counts())

print(df['rank'].value_counts())
```

```python
# Step 6: Histograms for numeric columns

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
    plt.figure()
    df[col].hist(bins=30)
    plt.title(f"Histogram of {col}")
    plt.savefig(f"hist_{col}.png")
    plt.show()
```

```python
# Step 7: Box plots for at least 2 columns
plt.figure()
sns.boxplot(data=df[numeric_cols[:2]])
plt.title("Box Plot of First Two Numeric Columns")
plt.savefig("boxplot.png")
plt.show()
```

```python
# Step 8: Pairplot for numeric columns
sns.pairplot(df[numeric_cols])
plt.savefig("pairplot.png")
plt.show()
```
