```python
# Multi-Chart Visualization Notebook

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset (Titanic as example)

df = sns.load_dataset("titanic")

# 1. Bar Chart (Matplotlib)

plt.figure(figsize=(6,4))
df['class'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Passenger Count by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
print("Insight: Most passengers traveled in Third Class.")
```

```python
# 2. Line Chart (Matplotlib)

plt.figure(figsize=(6,4))
df.groupby("age")["fare"].mean().plot(kind="line", color="green")
plt.title("Average Fare by Age")
plt.xlabel("Age (years)")
plt.ylabel("Fare ($)")
plt.show()
print("Insight: Younger passengers tended to pay lower fares.")
```

```python
# 3. Histogram (Matplotlib)

plt.figure(figsize=(6,4))
plt.hist(df["age"].dropna(), bins=20, color="orange", edgecolor="black")
plt.title("Age Distribution of Passengers")
plt.xlabel("Age (years)")
plt.ylabel("Frequency")
plt.show()
print("Insight: Most passengers were between 20 and 40 years old.")
```

```python
# 4. Scatter Plot (Matplotlib)

plt.figure(figsize=(6,4))
plt.scatter(df["age"], df["fare"], alpha=0.5)
plt.title("Fare vs Age")
plt.xlabel("Age (years)")
plt.ylabel("Fare ($)")
plt.show()
print("Insight: Higher fares were paid by a few older passengers.")
```

```python
# 5. Box Plot (Seaborn)

plt.figure(figsize=(6,4))
sns.boxplot(x="class", y="fare", data=df)
plt.title("Fare Distribution by Class")
plt.xlabel("Class")
plt.ylabel("Fare ($)")
plt.show()
print("Insight: First Class fares had the widest spread and highest median.")
```

```python
# 6. Heatmap (Seaborn)

plt.figure(figsize=(8,6))
corr = df[["age","fare","survived"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
print("Insight: Fare shows a positive correlation with survival.")
```
