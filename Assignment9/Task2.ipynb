```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset("titanic")

print("This dashboard explores passenger demographics and survival patterns on the Titanic.")
print("It highlights how age, fare, class, and gender influenced passenger experiences.")
print("Together, the four charts tell a cohesive story about customer behavior and survival outcomes.")
```

```python
# Set consistent style and palette

sns.set_style("whitegrid")
palette = "Set2"


# Create 2x2 subplot grid

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
```

```python
# Panel 1: Histogram of Age

sns.histplot(df["age"].dropna(), bins=20, kde=True, color="skyblue", ax=axes[0,0])
axes[0,0].set_title("Age Distribution of Passengers")
axes[0,0].set_xlabel("Age (years)")
axes[0,0].set_ylabel("Count")
```

```python
# Panel 2: Boxplot of Fare by Class

sns.boxplot(x="class", y="fare", data=df, hue="class", palette=palette, legend=False, ax=axes[0,1])

axes[0,1].set_title("Fare Distribution by Class")
axes[0,1].set_xlabel("Class")
axes[0,1].set_ylabel("Fare ($)")
```

```python
# Panel 3: Bar chart of Survival by Sex

sns.barplot(x="sex", y="survived", data=df, hue="sex", palette=palette, legend=False, ax=axes[1,0])
axes[1,0].set_title("Survival Rate by Gender")
axes[1,0].set_xlabel("Gender")
axes[1,0].set_ylabel("Survival Rate")
```

```python
# Panel 4: Scatter plot of Age vs Fare colored by Survival

sns.scatterplot(x="age", y="fare", hue="survived", data=df, palette=palette, ax=axes[1,1])
axes[1,1].set_title("Age vs Fare Colored by Survival")
axes[1,1].set_xlabel("Age (years)")
axes[1,1].set_ylabel("Fare ($)")
```

```python
# Adjust layout

plt.tight_layout()

# Export at 300 DPI

plt.savefig("storytelling_dashboard.png", dpi=300)
plt.show()
```
