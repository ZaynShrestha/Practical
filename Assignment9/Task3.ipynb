```python
import plotly.express as px
import seaborn as sns
import pandas as pd

# Load Titanic dataset (you can replace with your own ETL dataset)

df = sns.load_dataset("titanic")

# 1. Interactive Bar Chart
fig_bar = px.bar(
    df,
    x="class",
    color="sex",
    hover_data=["age", "fare"],
    title="Passenger Count by Class (Interactive)"
)
fig_bar.write_html("bar_chart.html")
fig_bar.show()
```

```python
# 2. Interactive Line Chart

avg_fare_by_age = df.groupby("age", as_index=False)["fare"].mean()
fig_line = px.line(
    avg_fare_by_age,
    x="age",
    y="fare",
    color_discrete_sequence=["green"],
    hover_data=["fare"],
    title="Average Fare by Age (Interactive)"
)
fig_line.write_html("line_chart.html")
fig_line.show()
```

```python
# 3. Interactive Scatter Plot

fig_scatter = px.scatter(
    df,
    x="age",
    y="fare",
    color="class",
    hover_data=["sex", "survived"],
    title="Fare vs Age (Interactive)"
)
fig_scatter.write_html("scatter_chart.html")
fig_scatter.show()
```

```python
# 4. New Chart Type: Sunburst

# Design choice: Sunburst shows hierarchical breakdown of survival by class and sex.

df["class"] = df["class"].astype(str)
df["sex"] = df["sex"].astype(str)
df["survived"] = df["survived"].astype(str)


fig_sunburst = px.sunburst(
    df,
    path=["class", "sex", "survived"],
    values="fare",
    color="class",
    hover_data=["age", "fare"],
    title="Sunburst of Survival by Class and Sex"
)
fig_sunburst.write_html("sunburst_chart.html")
fig_sunburst.show()
```
